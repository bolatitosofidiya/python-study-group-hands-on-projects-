# Excercise 1

```python
# Patient Registration System (Healthcare Industry) using the .format() method

first_name = "Grace"
last_name = "Johnson"
age = 34
blood_group = "O+"
department = "Cardiology"

print("""
========== CAREPLUS HOSPITAL ==========
Patient Name: {}
Age: {} Years
Blood Group: {}
Department: {}
=======================================
""".format(first_name.upper() + " " + last_name.upper(), age, blood_group, department))
```

```python
# Patient Registration System (Healthcare Industry) using the f-string method

print(f"""
========== CAREPLUS HOSPITAL ==========
Patient Name: {first_name.upper() + " " + last_name.upper()}
Age: {age} Years
Blood Group: {blood_group}
Department: { department}
=======================================
""")
```

# Excercise 2

```python
# Fraud Detection Log Analyzer (Banking & Cybersecurity Industry)

message = "     LOGIN SUCCESSFUL FROM ATM-045 ON 2026-07-25     "

print(message)
clean_message = (message.strip()) # rmeove leading and trailing spaces
print(message) # display cleaned message
print(clean_message.lower()) # convert all letters to lowercase or small letter
print(clean_message.find("SUCCESSFUL")) # find the start postion of the word successful
print(clean_message.replace("SUCCESSFUL","VERIFIED")) # replace the word succesful with verified
print(clean_message)
print(len(message)) # Total number of characters before cleaning
print(len(clean_message)) # Total number of characters after cleaning
print(len(clean_message.split())) # Total number of words in the cleaned log
```

# Excerise 3

```python
# Movie Recommendation Search Engine (Entertainment & AI Industry

movie_title = input("Enter Movie Name: ")

print("First Character: ",movie_title[0]) # print the first character
print("Last Character: ", movie_title[-1]) # print the last character
print("First Word: ",movie_title.split()[0]) # print the first word
print("Last Word: ",movie_title.split()[-1]) # print the last word
print('Position of "Returns":',movie_title.find("Returns"))
print("Split Result: ",movie_title.split())

print(len(movie_title.split())) # Number of words
print(len(movie_title)) # Number of characters
print(movie_title.upper()) # convert to uppercase
```
