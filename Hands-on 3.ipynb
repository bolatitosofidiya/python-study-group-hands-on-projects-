```python
# . Smart Shopping Discount Calculator 

laptop_bag = 18500
wireless_mouse = 12000
keyboard = 24500
usb_drive = 5000

total_purchase = laptop_bag + wireless_mouse + keyboard + usb_drive

if total_purchase >= 50000:
    discount_qualify = True
    discount_amount =int( total_purchase * 0.10)

else:
     discount_qualify = False
     discount_amount = 0

print("Total Purchase: ",total_purchase)
print("Discount Qualified: ",total_purchase >= 50000)
print("Discount Amount: ",discount_amount)
print("Final Amount: ",total_purchase - discount_amount)
```
