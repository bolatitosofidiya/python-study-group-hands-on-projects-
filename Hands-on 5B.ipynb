# Exercise 2

```python
# Payroll System

employee_name = "David Musa"
department = "Finance"
basic_salary = 185000
bonus = 35000
tax_rate = 7.5

gross_salary = basic_salary + bonus
tax = tax_rate / 100 * (basic_salary + bonus)
net_salary = gross_salary - tax


print(f"""
==================================
        PAYROLL SUMMARY
==================================

Employee : {employee_name}

Department : {department}

Basic Salary : ₦{basic_salary:.2f}

Bonus : ₦{bonus:.02f}

Gross Salary : ₦{gross_salary:.2f}

Tax (7.5%) : ₦{tax:.2f}

Net Salary : ₦{net_salary:.2f}

==================================

""")
```

# Exercise 3

```python
# E-commerce Order Receipt

customer_name ="Amina Bello"
product = "Wireless Headphones"
quantity = 3
unit_price = 27500
total_cost = unit_price * quantity

if quantity >= 2:
    discount= total_cost * 0.10
    
else:
    discount = 0

print(f"""
====================================

      ONLINE STORE RECEIPT

====================================

Customer : {customer_name}

Product : {product}

Quantity : {quantity}

Unit Price : ₦{unit_price:.2f}

Total Cost : ₦{total_cost:.2f}

Discount  :  ₦{discount:.2f}

Final Amount : ₦{total_cost - discount:.2f}

Thank you for shopping with us!

====================================
""")
```
