# Exercise 3

```python
# Inventory Management System 

stock = int(input("Initial stock: "))
sold_stock = int(input("Items sold: "))
new_shipment = int(input("New items: "))
full_carton = int(input("Full cartons: "))

stock -= sold_stock
print("Remaining stock after sale: ",stock)

stock += new_shipment
print("Updated stock after restocking: ",stock)

stock //= full_carton
print("Full cartons: ",stock)

stock %= full_carton
print("Remaining stock: ",stock)
```
